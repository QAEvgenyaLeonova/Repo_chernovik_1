```python
#Пример: повторение текста
def repeat(text, times = 1):
    return text * times

print(repeat('hi'))
print(repeat('hi', 3))

```

```python
#Пример: склейка слов с разделителем
def join_words(words1, words2, sep=''):
    return words1 + sep + words2

print(join_words('King', 'Road'))
print(join_words("Dragon", "stone", "-"))

```

```python
#Пример: несколько параметров по умолчанию
def make_l(sumbool='-', lenght=10):
    return sumbool * lenght

print(make_l())
print(make_l('*'))
print(make_l('*', 5))
print(make_l('#', 3))
```

```python
def get_hidden_card(card_number, stars_count=4):
    visible_digits_line = card_number[-4:]
    return f"{'*' * stars_count}{visible_digits_line}"

print(get_hidden_card('1234567812345678'))
print(get_hidden_card('2034399002125581', 6))    # => ******5581
print(get_hidden_card('9876543210987654', 2))    # => **7654
print(get_hidden_card('1111222233334444', 0))
```

```python
#Позиционный вызов: по порядку, как в списке
repeat("Hi", 3)  # "Hi" — первый, 3 — второй

#Именованный вызов: подписываем каждый ингредиент
repeat(text="Hi", times=3)  # явно сказали, что есть что
repeat(times=3, text="Hi")  # порядок поменяли — а результат тот же!

'''Позиционные аргументы — передаём значения по порядку. Важно не перепутать последовательность.

Именованные аргументы — явно указываем, какое значение какому параметру соответствует. Порядок не важен, удобно, когда параметров много или нужно изменить только один.

Можно комбинировать: сначала позиционные (если есть), потом именованные.'''

```

```python
def make_line(symbol="-", length=10):
    return symbol * length

make_line()              # => ---------- (всё по умолчанию)
make_line(length=5)     # => ----- (меняем только длину, символ остался дефисом)
make_line(symbol="*")   # => ********** (меняем только символ, длина осталась 10)
make_line("*", 5)     # => ***** (позиционно: сначала символ, потом длина)
make_line(length=3, symbol="#")  # => ### (именованно, порядок не важен!)

```

```python
def trim_and_repeat(text, offset=2, repetitions=1):
    return text[offset:] * repetitions

result = trim_and_repeat("Hello, world!", offset=2, repetitions=3)
print(result)
```
